```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

import warnings
warnings.filterwarnings('ignore')
```

```python
try :
    df = pd.read_csv ("dataset/sales1.csv")
    print ("Data berhasil masuk")
except FileNotFoundError :
    print ("Data tidak masuk")
```

```python
df['date'] = pd.to_datetime(df['date'], dayfirst=True)
print(df.dtypes)
```

jadi untuk ngubuah date jadi date time itu pakai pd.to datetime, dan pakai dayfirst=True karena kita hari duluan, bukan Month duluan.

Group data berdasarkan date (pakai groupby),Karena ada kolom sales dan quantity kita pakai sum().

```python
#Ini contoh kita pakai group by untuk Total sales per tanggal
grouped_date1 = df.groupby('date')['sales'].sum()
print(grouped_date1)
```

```python
#Ini contoh kita pakai group by untuk Total quantity & sales per tanggal
grouped_date2 = df.groupby('date')[['quantity', 'sales']].sum()
print(grouped_date2)
```

```python

#group data berdasarkan date nih
grouped_date = df.groupby(['date', 'store', 'product']) [['quantity', 'sales']].sum()
#kita simpan ke data frame
grouped_date = grouped_date.reset_index()
grouped_date.head(10)
```

Pada tahap ini, data penjualan dikelompokkan berdasarkan tanggal (date) untuk memperoleh total jumlah produk (quantity) dan total penjualan (sales) per hari.
